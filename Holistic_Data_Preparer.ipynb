# Part A: Conceptual Foundation

## 1. What is Data Analysis?

Data Analysis is the process of collecting, cleaning, transforming, and interpreting data to extract meaningful insights and support decision-making. It helps convert raw data into useful information that can guide business strategies and predictions.

### Key Steps in Data Analysis:
- **Data Collection**: Gathering raw data from sources like CSV files, databases, APIs, etc.
- **Data Cleaning**: Handling missing values, removing duplicates, correcting errors.
- **Data Transformation**: Converting data into a structured and usable format.
- **Exploratory Data Analysis (EDA)**: Identifying patterns, trends, and relationships.
- **Data Visualization**: Representing data using charts and graphs.
- **Interpretation**: Drawing conclusions from the data.

**Goal:** Transform raw data into actionable insights.

---

## 2. How to Plan a Data Science Project

A Data Science project follows a structured lifecycle:

### Steps:

1. **Problem Understanding**  
   Clearly define the objective.  
   Example: Predict whether a customer will default on a loan.

2. **Data Collection**  
   Gather data from sources like CSV, JSON, SQL databases, or APIs.

3. **Data Preprocessing**  
   - Handle missing values  
   - Remove outliers  
   - Encode categorical variables  
   - Normalize or scale features  

4. **Exploratory Data Analysis (EDA)**  
   Analyze distributions, correlations, and patterns.

5. **Feature Engineering**  
   Create or modify features to improve model performance.

6. **Model Building**  
   Select and train machine learning algorithms.

7. **Model Evaluation**  
   Use metrics such as accuracy, precision, recall, and F1-score.

8. **Deployment (Optional)**  
   Deploy the model in real-world applications.

---

## 3. How to Frame a Machine Learning Problem

Framing a machine learning problem involves defining what needs to be predicted and how.

### Steps:

- **Define Target Variable**  
  Example: `default_flag` (0 = No Default, 1 = Default)

- **Identify Problem Type**  
  - Classification (predict categories)  
  - Regression (predict numerical values)  
  - Clustering (group data)  

  This project is a **Binary Classification Problem**.

- **Select Input Features**  
  Example: age, income, credit score, etc.

- **Define Evaluation Metrics**  
  Accuracy, Precision, Recall, F1-score.

- **Understand Constraints**  
  Data quality, imbalance, and real-world impact.

---

## 4. Tensors and NumPy Explanation

### What is a Tensor?

A Tensor is a multi-dimensional array used to represent data.

### Types of Tensors:
- **Scalar (0D)**: Single value  
- **Vector (1D)**: List of values  
- **Matrix (2D)**: Table of values

## 6)Handling Missing Data

### create a clean copy of dataset

```python
# create a clean copy of dataset
df_clean = df.copy()

# 6) handle missing data using simple imputer for numerical features

from sklearn.impute import SimpleImputer

print("\nHandling Missing Values for Numerical Columns using Mean and Median")

# select numerical columns
num_cols = df_clean.select_dtypes(include=["int64", "float64"]).columns

# mean imputation
mean_imputer = SimpleImputer(strategy="mean")
df_clean[num_cols] = mean_imputer.fit_transform(df_clean[num_cols])

print("\nMissing Values After Mean Imputation")
print(df_clean[num_cols].isnull().sum())
```

### a)Simple imputer 

```python
# simple imputer for categorical features using most frequent

print("\nHandling Missing Values for Categorical Columns using Most Frequent")

cat_cols = df_clean.select_dtypes(include=["object"]).columns

cat_imputer = SimpleImputer(strategy="most_frequent")
df_clean[cat_cols] = cat_imputer.fit_transform(df_clean[cat_cols])

print("\nMissing Values After Categorical Imputation")
print(df_clean[cat_cols].isnull().sum())
```

### b)missing indicator and random sample imputation

```python
# missing indicator and random sample imputation for annual_income

print("\nApplying Missing Indicator and Random Sample Imputation on annual_income")

# create missing indicator
df_clean["annual_income_missing"] = df_clean["annual_income"].isnull().astype(int)

# random sample imputation
random_sample = df_clean["annual_income"].dropna().sample(df_clean["annual_income"].isnull().sum(), random_state=42)

df_clean.loc[df_clean["annual_income"].isnull(), "annual_income"] = random_sample.values

print("\nMissing Values After Random Sample Imputation")
print(df_clean["annual_income"].isnull().sum())
```

### c)knn imputer

```python
# knn imputer for multivariate imputation

from sklearn.impute import KNNImputer

print("\nApplying KNN Imputer on Numerical Features")

num_cols = df_clean.select_dtypes(include=["int64", "float64"]).columns

knn_imputer = KNNImputer(n_neighbors=5)

df_clean[num_cols] = knn_imputer.fit_transform(df_clean[num_cols])

print("\nMissing Values After KNN Imputation")
print(pd.DataFrame(df_clean[num_cols]).isnull().sum())
```

### d) MICE imputation

```python
# mice imputation using iterative imputer

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

print("\nApplying MICE Imputation")

num_cols = df_clean.select_dtypes(include=["int64", "float64"]).columns

mice_imputer = IterativeImputer(random_state=42)

df_clean[num_cols] = mice_imputer.fit_transform(df_clean[num_cols])

print("\nMissing Values After MICE Imputation")
print(pd.DataFrame(df_clean[num_cols]).isnull().sum())
```

# Part D: Outlier Handling

## 7) Detect And Handle Outliers

### a)Z-score

```python
# create working copy
df_outlier = df_clean.copy()

print("\nApplying Z Score Method for Outlier Detection and Treatment")

from scipy import stats
import numpy as np

cols = ["annual_income", "loan_amount", "credit_score"]

print("\nShape Before Z Score Treatment")
print(df_outlier.shape)

# count outliers before
for col in cols:
    z = np.abs(stats.zscore(df_outlier[col]))
    print(f"\nOutliers in {col} before treatment:", (z > 3).sum())

# apply capping
for col in cols:
    z = np.abs(stats.zscore(df_outlier[col]))
    df_outlier[col] = np.where(z > 3, df_outlier[col].median(), df_outlier[col])

print("\nShape After Z Score Treatment")
print(df_outlier.shape)

# verify outliers after
for col in cols:
    z = np.abs(stats.zscore(df_outlier[col]))
    print(f"\nOutliers in {col} after treatment:", (z > 3).sum())
```

### b) IQR

```python
# create working copy
df_iqr = df_clean.copy()

print("\nApplying IQR Method for Outlier Detection and Removal")

# select important numerical columns
cols = ["annual_income", "loan_amount", "credit_score"]

# shape before treatment
print("\nShape Before IQR Treatment")
print(df_iqr.shape)

# remove outliers using IQR
for col in cols:
    Q1 = df_iqr[col].quantile(0.25)
    Q3 = df_iqr[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df_iqr = df_iqr[(df_iqr[col] >= lower_bound) & (df_iqr[col] <= upper_bound)]

# shape after treatment
print("\nShape After IQR Treatment")
print(df_iqr.shape)

# summary after treatment
print("\nSummary After IQR Treatment")
print(df_iqr[cols].describe())
```

### c) Percentile Method

```python
# create working copy
df_percentile = df_clean.copy()

print("\nApplying Percentile Method for Outlier Treatment")

import numpy as np

# select important numerical columns
cols = ["annual_income", "loan_amount", "credit_score"]

# shape before treatment
print("\nShape Before Percentile Treatment")
print(df_percentile.shape)

# count extreme values before
for col in cols:
    lower = df_percentile[col].quantile(0.01)
    upper = df_percentile[col].quantile(0.99)
    count = ((df_percentile[col] < lower) | (df_percentile[col] > upper)).sum()
    print(f"\nExtreme values in {col} before treatment:", count)

# apply capping using percentile
for col in cols:
    lower = df_percentile[col].quantile(0.01)
    upper = df_percentile[col].quantile(0.99)

    df_percentile[col] = np.where(df_percentile[col] < lower, lower, df_percentile[col])
    df_percentile[col] = np.where(df_percentile[col] > upper, upper, df_percentile[col])

# shape after treatment
print("\nShape After Percentile Treatment")
print(df_percentile.shape)

# verify extreme values after
for col in cols:
    lower = df_percentile[col].quantile(0.01)
    upper = df_percentile[col].quantile(0.99)
    count = ((df_percentile[col] < lower) | (df_percentile[col] > upper)).sum()
    print(f"\nExtreme values in {col} after treatment:", count)

# summary after treatment
print("\nSummary After Percentile Treatment")
print(df_percentile[cols].describe())
```

### d) Winsorization 

```python
# create working copy
df_winsor = df_clean.copy()

print("\nApplying Winsorization Technique for Outlier Treatment")

from scipy.stats.mstats import winsorize
import numpy as np

cols = ["annual_income", "loan_amount", "credit_score"]

print("\nShape Before Winsorization")
print(df_winsor.shape)

# store original thresholds
bounds = {}

for col in cols:
    lower = df_winsor[col].quantile(0.01)
    upper = df_winsor[col].quantile(0.99)
    bounds[col] = (lower, upper)

    count = ((df_winsor[col] < lower) | (df_winsor[col] > upper)).sum()
    print(f"\nExtreme values in {col} before treatment:", count)

# apply winsorization and convert to normal array
for col in cols:
    df_winsor[col] = np.array(winsorize(df_winsor[col], limits=[0.01, 0.01]))

print("\nShape After Winsorization")
print(df_winsor.shape)

# verify using SAME old thresholds
for col in cols:
    lower, upper = bounds[col]
    count = ((df_winsor[col] < lower) | (df_winsor[col] > upper)).sum()
    print(f"\nExtreme values in {col} after treatment:", count)

print("\nSummary After Winsorization")
print(df_winsor[cols].describe())
```

# Part E: Feature Engineering

## 8) handle variable types

```python
# create working copy
df_feat = df_clean.copy()

# 8) handle variable types

print("\nHandling Date and Mixed Variables")

# convert join_date to datetime
df_feat["join_date"] = pd.to_datetime(df_feat["join_date"])

# extract date features
df_feat["join_year"] = df_feat["join_date"].dt.year
df_feat["join_month"] = df_feat["join_date"].dt.month
df_feat["join_day"] = df_feat["join_date"].dt.day
df_feat["join_weekday"] = df_feat["join_date"].dt.weekday

print("\nNew Date Features Created")
print(df_feat[["join_year", "join_month", "join_day", "join_weekday"]].head())
```

## 9) encoding categorical variables

```python
# 9) encoding categorical variables

# ordinal encoding for education_level
print("\nApplying Ordinal Encoding on education_level")

education_order = {
    "Primary": 1,
    "Secondary": 2,
    "Graduate": 3,
    "Post-Graduate": 4
}

df_feat["education_level"] = df_feat["education_level"].map(education_order)

print("\nEncoded education_level")
print(df_feat["education_level"].value_counts())


# label encoding for gender
print("\nApplying Label Encoding on gender")

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df_feat["gender"] = le.fit_transform(df_feat["gender"])

print("\nEncoded gender values")
print(df_feat["gender"].value_counts())


# one hot encoding for region and loan_purpose
print("\nApplying One Hot Encoding on region and loan_purpose")

df_feat = pd.get_dummies(df_feat, columns=["region", "loan_purpose"], drop_first=True)

print("\nColumns After One Hot Encoding")
print(df_feat.columns)
```

## 10) encoding numerical features

```python
# 10) encoding numerical features

# binning for annual_income
print("\nApplying Binning on annual_income")

df_feat["income_bin"] = pd.cut(
    df_feat["annual_income"],
    bins=4,
    labels=["Low", "Medium", "High", "Very High"]
)

print("\nIncome Bins Created")
print(df_feat["income_bin"].value_counts())


# binarization for credit_score
print("\nApplying Binarization on credit_score")

df_feat["credit_flag"] = (df_feat["credit_score"] > 700).astype(int)

print("\nCredit Flag Distribution")
print(df_feat["credit_flag"].value_counts())


# quantile binning for transaction_count
print("\nApplying Quantile Binning on transaction_count")

df_feat["transaction_bin"] = pd.qcut(
    df_feat["transaction_count"],
    q=4,
    labels=["Low", "Medium", "High", "Very High"]
)

print("\nTransaction Bins")
print(df_feat["transaction_bin"].value_counts())


# k means binning for transaction_count
print("\nApplying K Means Binning on transaction_count")

from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=4, random_state=42)

df_feat["transaction_cluster"] = kmeans.fit_predict(
    df_feat[["transaction_count"]]
)

print("\nTransaction Clusters Created")
print(df_feat["transaction_cluster"].value_counts())
```

# Part G: Feature Construction & Transformation

## 12 Apply Transformation

```python
# 12) apply function transformer transformations

from sklearn.preprocessing import FunctionTransformer
import numpy as np

print("\nApplying Function Transformer Transformations")

df_trans = df_feat.copy()

# select columns
cols = ["annual_income", "loan_amount", "spending_ratio"]

# log transform
log_transformer = FunctionTransformer(np.log1p)
df_trans[cols] = log_transformer.fit_transform(df_trans[cols])

print("\nLog Transformation Applied")
print(df_trans[cols].head())
```

```python
# 12) apply reciprocal and square root transformations

print("\nApplying Reciprocal and Square Root Transformations")

df_trans2 = df_feat.copy()

# reciprocal transform
df_trans2["loan_amount_reciprocal"] = 1 / (df_trans2["loan_amount"] + 1)

# square root transform
df_trans2["income_sqrt"] = np.sqrt(df_trans2["annual_income"])

print("\nNew Transformed Columns Created")
print(df_trans2[["loan_amount_reciprocal", "income_sqrt"]].head())
```

```python
# 12) apply power transformer (yeo johnson and box cox)

from sklearn.preprocessing import PowerTransformer

print("\nApplying Power Transformer")

df_power = df_feat.copy()

cols = ["annual_income", "loan_amount"]

# yeo johnson
pt = PowerTransformer(method="yeo-johnson")
df_power[cols] = pt.fit_transform(df_power[cols])

print("\nYeo Johnson Transformation Applied")
print(df_power[cols].describe())
```

```python
# 12) apply column transformer

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

print("\nApplying Column Transformer")

df_ct = df_feat.copy()

num_cols = ["annual_income", "loan_amount", "credit_score"]
cat_cols = ["employment_type"]

ct = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), num_cols),
        ("cat", OneHotEncoder(drop="first"), cat_cols)
    ]
)

transformed_data = ct.fit_transform(df_ct)

print("\nColumn Transformer Applied")
print("\nTransformed Data Shape")
print(transformed_data.shape)
```

```python
# 13) construct new features

print("\nConstructing New Features")

df_construct = df_feat.copy()

# debt to income ratio
df_construct["debt_to_income"] = df_construct["loan_amount"] / df_construct["annual_income"]

# average monthly transactions
df_construct["avg_monthly_txn"] = df_construct["transaction_count"] / 6

# spending to income ratio already exists but ensure consistency
df_construct["spending_income_ratio"] = df_construct["spending_ratio"]

print("\nNew Features Created")
print(df_construct[["debt_to_income", "avg_monthly_txn", "spending_income_ratio"]].head())
```
